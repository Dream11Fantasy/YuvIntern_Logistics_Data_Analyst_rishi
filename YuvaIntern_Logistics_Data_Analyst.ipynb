```python
import pandas as pd
import numpy as np

data = {
    "Shipment_ID": [1, 2, 3, 4, 5],
    "Distance_km": [120, 250, 180, 400, 150],
    "Shipment_Quantity": [20, 35, 25, 50, 30],
    "Transport_Cost": [1800, 3500, 2500, 5500, 2100],
    "Delivery_Time_Days": [2, 4, 3, 6, 3]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Basic analysis of the logistics data

print("Number of shipments:", len(df))
print("Average distance:", df["Distance_km"].mean(), "km")
print("Average shipment quantity:", df["Shipment_Quantity"].mean())
print("Average transport cost:", df["Transport_Cost"].mean())
print("Average delivery time:", df["Delivery_Time_Days"].mean(), "days")
```
